```python
import pandas as pd
import numpy as np
```

```python
df_amazon = pd.read_csv('../data/raw/amazon_delivery.csv')
```

```python
df_amazon.head()
```

```python
df_amazon['Delivery_Time'].mean()
```

```python
absolut_error = (125 - df_amazon['Delivery_Time']).abs()
mae = absolut_error.mean()
mae
```

```python
absolut_error = ((125 - df_amazon['Delivery_Time']).abs())**2
RMSE = np.sqrt(absolut_error.mean())
RMSE
```

```python
df_amazon.query("Area == 'Metropolitian '")['Delivery_Time'].describe()
```

```python
df_metropolitian = df_amazon.query("Area == 'Metropolitian '")
```

```python
#The metropolitian MAE is:
absolut_error = (129.7 - df_metropolitian['Delivery_Time']).abs()
mae = absolut_error.mean()
mae
```

```python
#The metropolitian RMSE is:
absolut_error = ((129.7 - df_metropolitian['Delivery_Time']).abs())**2
RMSE = np.sqrt(absolut_error.mean())
RMSE
```
